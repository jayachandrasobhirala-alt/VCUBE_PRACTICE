MODERATE

```python
#9
s = "abc123xyz456"
print(re.findall(r'\d{3}', s))
```

```python
#10
s = "abc abc abc"
print(re.sub(r'abc', 'x', s, 2))
```

```python
#11
s = "a1b2c3"
print(re.split(r'\d', s))
```

```python
#12
s = "one1two2three3"
print(re.findall(r'[a-z]+\d', s))
```

```python
#13
s = "test@test.com"
print((re.match(r'\w+@\w+\.\w+', s)))
```

```python
#14
s = "cat scatter cater"
print(re.findall(r'cat', s))
```

```python
#15
s = "123 45 6789"
print(re.findall(r'\b\d{2}\b', s))
```

```python
#16
s = "hello\nworld"
print(re.findall(r'.', s))
```

```python
#17
s = "abc123"
print(re.search(r'\d+', s).group())
```
