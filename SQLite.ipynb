SQLITE


```python
import sqlite3
db="emp11.db"
conn=sqlite3.connect(db)
print("opened database successfully")
```

CREATING TABLE

```python
conn.execute('''
  create table company(
    id int primary key not null,
    name varchar(20) not null,
    age int not null,
    address varchar(50) not null,
    salary int not null
  )
''')
print("table is created")
conn.close()
```

INSERTING DATA INTO THE TABLE

```python
conn=sqlite3.connect(db)
conn.execute("insert into company(id,name,age,address,salary) values(01,'Jay',21,'Hyderabad',100000)")
conn.execute("insert into company(id,name,age,address,salary) values(02,'Jay2',21,'Hyderabad',200000)")
conn.execute("insert into company(id,name,age,address,salary) values(03,'Jay3',21,'Hyderabad',300000)")
conn.execute("insert into company(id,name,age,address,salary) values(04,'Jay4',21,'Hyderabad',400000)")
conn.commit()
print("data entered successfully")
conn.close()
```

DISPLAYING DATA

```python
conn=sqlite3.connect(db)
print("opened database")
dt=conn.execute("select id, name,age,address,salary from company")
for row in dt:
  print("id=",row[0])
  print("name=",row[1])
  print("age=",row[2])
  print("address=",row[3])
  print("salary=",row[4],"\n")
print("displayed successfully")
conn.close()
```

UPDATE

```python
conn=sqlite3.connect(db)
conn.execute("update company set salary=20000000 where id=01")
conn.commit()
print("updated successfully")
conn.close()
```
