```python
import re
```

1) Program to check valid phone numbers

```python
def validate_phonenumber(s):
  ans=bool(re.fullmatch(r'\d{10}',s))
  if(ans):
    print("phonenumber is valid")
  else:
    print("phonenumber is not valid")
lst=["1234567891","123","3453653"]
for i in lst:
  validate_phonenumber(i)
```

2. program to check valid name

```python
def valid_name(s):
  ans=bool(re.search(r'[A-Z][a-z]+',s))
  if(ans):
    print("name is valid")
  else:
    print("name is not valid")

lst=["JAY","Jay","jay"]
for i in lst:
  valid_name(i)
```

3. program to check valid password

```python
pattern = r'^(?=.*\d)(?=.*[@#$%]).{8,20}$'

passwords = ["weak", "strong1#", "nogood1", "excellent9$"]

for pwd in passwords:
    if re.fullmatch(pattern, pwd):
      print(f"'{pwd}' is valid")
    else:
      print(f"'{pwd}' is invalid")
```

4. valid email

```python
def valid_email(s):
  ans=bool(re.match(r'\w+@\w+\.\w+',s))
  if(ans):
    print("valid email")
  else:
    print("invalid email")
lst=["test@test.com","jayachandra","@test.com"]
for i in lst:
  valid_email(i)
```
