```python
import sqlite3
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()
```

```python
cursor.execute("""
CREATE TABLE students (
    student_id INTEGER PRIMARY KEY,
    name TEXT,
    dept_id INTEGER
)
""")

cursor.execute("""
CREATE TABLE departments (
    dept_id INTEGER PRIMARY KEY,
    dept_name TEXT
)
""")

cursor.execute("""
CREATE TABLE marks (
    mark_id INTEGER PRIMARY KEY,
    student_id INTEGER,
    subject TEXT,
    score INTEGER
)
""")
```

```python
cursor.executemany("INSERT INTO students VALUES (?, ?, ?)", [
    (1, 'Alice', 101),
    (2, 'Bob', 102),
    (3, 'Charlie', 101)
])

cursor.executemany("INSERT INTO departments VALUES (?, ?)", [
    (101, 'Computer Science'),
    (102, 'Mechanical')
])

cursor.executemany("INSERT INTO marks VALUES (?, ?, ?, ?)", [
    (1, 1, 'Math', 90),
    (2, 1, 'Science', 85),
    (3, 2, 'Math', 78),
    (4, 3, 'Science', 88)
])

conn.commit()
```

```python
#inner join
query = """
SELECT s.name, d.dept_name, m.subject, m.score
FROM students s
INNER JOIN departments d ON s.dept_id = d.dept_id
INNER JOIN marks m ON s.student_id = m.student_id
"""

result = cursor.execute(query).fetchall()
result
```

```python
#left join
query = """
SELECT s.name, d.dept_name, m.subject, m.score
FROM students s
LEFT JOIN departments d ON s.dept_id = d.dept_id
LEFT JOIN marks m ON s.student_id = m.student_id
"""

result = cursor.execute(query).fetchall()
result
```

```python
#crossjoin
query = """
SELECT s.name, m.subject
FROM students s
CROSS JOIN marks m
"""

result = cursor.execute(query).fetchall()
result
```
