expences over 2000


```python
lines = []
with open('expenses.txt', 'r') as f:
    for line in f:
        lines.append(line.strip())

print(lines[:100])
```

```python
exp=[]
for line in lines:
    exp.append(int(line))
print(exp[:100])
```

```python
greater_than_2k=[]
for i in exp:
  if(i>2000):
    greater_than_2k.append(i)
print(greater_than_2k)
```

words with "quot"

```python
words=[]
with open("wordlist.txt","r") as i:
  for word in i:
    words.append(word.strip())
print(words[:100])
```

```python
words_with_quot=[]
for i in words:
  if "quot" in i:
    words_with_quot.append(i)
print(words_with_quot[:100])
```

words with length>6 and first 3 = last 3

```python
for word in words:

  if len(word) >= 6 and word[:3] == word[-3:]:

    print(word, end=' ')
print()
```

```python
longest_word='a'
for word in words:
  if word[0]==word[-1] and len(word)>len(longest_word):
    longest_word=word
print(longest_word,end=' ')
```



Each line of the file population.txt contains a country name and its population, separated by a tab.

Write a program that uses the file to print out all the countries whose names start with G that have at least 500,000 people.


```python
lines = [line.strip() for line in open('population.txt')]
print(lines)
```

```python
for line in lines:
  parts = line.split() # Split by any whitespace

  # The population is always the last element, the country name is everything before it
  population = int(parts[-1])
  country = ' '.join(parts[:-1])

  if country[0] == 'G' and population >= 5000000:
    print(country, population)
```
