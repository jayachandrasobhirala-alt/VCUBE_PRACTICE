```python
import pandas as pd
df = pd.DataFrame({
    "num": [1, 2, 3, 4, 5]
})
df
```

```python
#squared
df["squared"]=df["num"].apply(lambda x:x**2)
df
```

```python
#even or odd
df["type"]=df["num"].apply(lambda x :"even" if x %2==0 else "odd" )
df
```

```python

df = pd.DataFrame({
    "A": [10, 20, 30],
    "B": [1, 2, 3]
})
df
```

```python
#sum of A,B
df["sum"]=df.apply(lambda row: row["A"]+row["B"],axis=1)
df
```

```python
df = pd.DataFrame({
    "name": ["A", "B", "C"],
    "marks": [85, 45, 72]
})
df
```

```python
#apply lambda for custom logic
df["grade"]=df["marks"].apply(
    lambda x : "A" if x > 80 else ("B" if x>60 else "C") 
    )
df
```

```python
df = pd.DataFrame({
    "age": [25, 17, 35],
    "salary": [50000, 20000, 80000]
})
df
```

```python
df["status"]=df.apply(
    lambda row: "eligible" if row["age"]>18 and row["salary"]>=50000 else "not eligible",axis=1
)
df
```

# Iterating over Dataframe

```python
#using iterrows()

df = pd.DataFrame({
    "name": ["A", "B", "C"],
    "marks": [85, 45, 72]
})
for index,rows in df.iterrows():
    print(index,rows["name"],rows["marks"])
```

```python
#itertuples()
for row in df.itertuples():
    print(row.name, row.marks)
```

```python
#iterate over columns
for col in df:
    print(col)
    print(df[col].values)
```

```python
#using items()
for col_name, col_data in df.items():
    print(col_name)
    print(col_data.tolist())
```
