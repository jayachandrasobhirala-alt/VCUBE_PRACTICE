what is broadcasting ? explain with an example apply functions like reshape ,resize, ndim on the output of broadcasting?


```python
import numpy as np
#ans:
# Definition:
# Broadcasting means NumPy automatically expands arrays of different shapes
# so mathematical operations can be performed without manually repeating data.
#example:
a = np.array([[1],[2],[3]])      # shape (3,1)
b = np.array([10,20,30])        # shape (3,)

c = a + b

print("Broadcasting Output:")
print(c)
```

```python
print("Shape:", c.shape)
print("ndim:", c.ndim)

# reshape
print(c.reshape(1,9))

# resize
c.resize((9,1))
print(c)
```

create an array apply all the aggregation functions like mean,median,mode,variance,std, dot product ?


```python
from scipy import stats

arr = np.array([10,20,30,40,50])

print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Mode:", stats.mode(arr))
print("Variance:", np.var(arr))
print("Standard Deviation:", np.std(arr))

# Dot Product
x = np.array([1,2,3])
y = np.array([4,5,6])

print("Dot Product:", np.dot(x,y))   # 1*4 +2*5 +3*6 = 32
```

explain the difference between flatten and ravel with an example ?

```python
mat = np.array([[1,2],[3,4]])

f = mat.flatten()
r = mat.ravel()

print("Flatten:", f)
print("Ravel:", r)

# Difference:
# flatten() returns copy
# ravel() returns view (linked to original)

r[0] = 100
print(mat)   # original changes
```

explain the diff between skewness,kurtosis?

```python
data = [2,3,4,5,6,7,100]

print("Skewness:", stats.skew(data))
print("Kurtosis:", stats.kurtosis(data))

# Skewness:
# Measures asymmetry of data

# >0 = right skewed
# <0 = left skewed
# =0 = symmetric

# Kurtosis:
# Measures peakness / tail heaviness

# High kurtosis = more outliers
# Low kurtosis = flatter distribution
```
